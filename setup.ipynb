```python
from pyspark.sql.functions import current_timestamp, lit, input_file_name

tables = ["customers", "data_dictionary", "exchange_rates","products","sales","stores"]

for tbl in tables:
    df = spark.table(f"retailer.azure_blob_storage.{tbl}") \
              .withColumn("_ingested_at", current_timestamp()) \
              .withColumn("_source_system", lit("fivetran"))
    df.write.mode("overwrite").saveAsTable(f"retailer.bronze.{tbl}_raw")
    print(f"Bronze: {tbl}_raw loaded with {df.count()} rows")
 
```
